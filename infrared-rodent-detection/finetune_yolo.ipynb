```python
from ultralytics import YOLO
import os
import random
```

```python
# Load a pretrained YOLO11n model
model = YOLO("models/yolo11n.pt")
```

```python
# Fine-tune the model on a custom dataset with multi-scale training
model.train(
    data="finetune.yaml",
    epochs=100,
    imgsz=640,
    multi_scale=True,  # enable multi-scale training
    batch=16,
    device=0,
    name="yolo11n-finetuned-ms"
    )
```

```python
# Evaluate the model's performance on the validation set
metrics = model.val(
    data="finetune.yaml",
    split="val",         
    imgsz=640,
    batch=16,
    device=0,
    name="yolo11n-finetuned-val-ms"
    )
```

```python
# Evaluate the model's performance on the test set
metrics_test = model.val(
    data="finetune.yaml",
    split="test",         
    imgsz=640,
    batch=16,
    device=0,
    name="yolo11n-finetuned-test-ms"
    )
```
